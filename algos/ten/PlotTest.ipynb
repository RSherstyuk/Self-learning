```python
import numpy as np
import matplotlib.pyplot as plt

# Параметры распределения
mu = 0      # среднее значение
sigma = 1   # стандартное отклонение

# Генерируем точки по оси X
x = np.linspace(-4, 4, 1000)

# Формула плотности нормального распределения
y = (1/(sigma * np.sqrt(2*np.pi))) * np.exp(-0.5*((x-mu)/sigma)**2)

# Строим график
plt.plot(x, y, 'b-', linewidth=2)
plt.title('Нормальное распределение N(0,1)')
plt.xlabel('x')
plt.ylabel('Плотность вероятности')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Параметры
mu, sigma = 0, 1
x = np.linspace(-4, 4, 1000)

# Используем готовую функцию плотности
y = norm.pdf(x, mu, sigma)

plt.plot(x, y, 'r-', linewidth=2, label=f'N({mu},{sigma}²)')
plt.title('Нормальное распределение')
plt.xlabel('x')
plt.ylabel('PDF')
plt.legend()
plt.grid(True)
plt.show()
```
